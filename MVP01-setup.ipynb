#MVP01 - Setup do ambiente
Implementação da estrutura do Lakehouse Catálogo MVP, com a definição dos schemas das camadas e a configuração do volume de staging.

```python
%sql
DROP CATALOG IF EXISTS mvp CASCADE
```

```python
%sql
CREATE CATALOG IF NOT EXISTS mvp
COMMENT  'MVP de Engenharia de Dados — pipeline sobre o dataset Olist (Kaggle) '
```

```python
%sql
USE CATALOG mvp
```

```python
%sql
DROP SCHEMA IF EXISTS staging CASCADE
```

```python
%sql
CREATE SCHEMA staging
COMMENT 'Arquivos originais, sem processamento'
```

```python
%sql
DROP SCHEMA IF EXISTS bronce CASCADE
```

```python
%sql
CREATE SCHEMA bronce
COMMENT 'Camada Bronze: cópia fiel dos arquivos em Delta.'
```

```python
%sql
DROP SCHEMA IF EXISTS silver CASCADE
```

```python
%sql
CREATE SCHEMA silver
COMMENT 'Camada Silver: dados tipados, padronizados e deduplicados.'
```

```python
%sql
DROP SCHEMA IF EXISTS gold CASCADE
```

```python
%sql
CREATE SCHEMA gold
COMMENT 'Camada Gold: modelo dimensional para consumo analítico.';
```

```python
%sql
CREATE VOLUME IF NOT EXISTS mvp.staging.kaggle_raw
COMMENT 'Arquivos CSV originais do dataset Olist baixados do Kaggle. Nenhuma transformação aplicada.'
```

```python
%sql
SELECT schema_name, comment
FROM mvp.information_schema.schemata
WHERE schema_name IN ('staging','bronce','silver','gold')
ORDER BY schema_name;
```
